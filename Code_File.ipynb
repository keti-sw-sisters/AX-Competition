## 기상청 (대전, 청주, 보은)

```python
import pandas as pd
import numpy as np

# 파일 읽기
df = pd.read_csv("기상청 자료 (대전, 충주, 보은).csv",encoding="cp949")

df
```

```python
df['일시'] = pd.to_datetime(df['일시'])
df['일강수량(mm)'] = df['일강수량(mm)'].fillna(0)
```

```python
numeric_cols = [
    '평균기온(°C)', '최저기온(°C)', '최고기온(°C)',
    '일강수량(mm)', '평균 풍속(m/s)',
    '평균 상대습도(%)', '합계 일조시간(hr)',
    '합계 일사량(MJ/m2)', '평균 전운량(1/10)'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
weather_daily = (
    df
    .groupby('일시')
    .agg({
        '평균기온(°C)': 'mean',
        '최저기온(°C)': 'mean',
        '최고기온(°C)': 'mean',
        '일강수량(mm)': 'mean',
        '평균 풍속(m/s)': 'mean',
        '평균 상대습도(%)': 'mean',
        '합계 일조시간(hr)': 'mean',
        '합계 일사량(MJ/m2)': 'mean',
        '평균 전운량(1/10)': 'mean'
    })
    .reset_index()
    .sort_values('일시')
)
#weather_daily = weather_daily.iloc[:-1]

print(weather_daily.shape)
weather_daily.tail()

```

## 수자원공사_조류데이터

```python
# 파일 읽기
df = pd.read_excel("댐 조류 모니터링.xlsx")
df.columns = df.columns.get_level_values(0)
df = df.reset_index(drop=True)
df["조사일"] = pd.to_datetime(df["조사일"])

df.head()
```

```python
# 2. 컬럼 이름 수정
df.columns = df.columns.str.replace("\n", "").str.strip()

# 컬럼 rename (핵심만)
df = df.rename(columns={
    "유해남조류 세포수 (cells/㎖)(1+2+3+4)": "total_cyano",
    "(1)Microcystis": "microcystis",
    "(2)Anabaena": "anabaena",
    "(3)Oscillatoria": "oscillatoria",
    "(4)Aphanizomenon": "aphanizomenon"
})

# 날짜 변환
df["조사일"] = pd.to_datetime(df["조사일"])

# 정렬
df = df.sort_values(["채수위치", "조사일"])
```

```python
# 3. 주 → 일 변환 (growth)
def expand_to_daily(df, target_cols):
    results = []

    for site, group in df.groupby("채수위치"):
        group = group.sort_values("조사일").reset_index(drop=True)

        for i in range(len(group) - 1):
            row_start = group.iloc[i]
            row_end = group.iloc[i + 1]

            start = row_start["조사일"]
            end = row_end["조사일"]
            days = (end - start).days

            date_range = pd.date_range(start=start, end=end, freq="D")

            temp = pd.DataFrame({"조사일": date_range})
            temp["채수위치"] = site

            for col in target_cols:
                y0 = row_start[col]
                y1 = row_end[col]

                eps = 1e-6
                y0, y1 = y0 + eps, y1 + eps

                r = np.log(y1 / y0) / days if days > 0 else 0
                temp[col] = [y0 * np.exp(r * t) for t in range(days + 1)]

            results.append(temp.iloc[:-1])

        results.append(group.iloc[-1:])

    return pd.concat(results).reset_index(drop=True)


target_cols = ["total_cyano", "microcystis", "anabaena", "oscillatoria", "aphanizomenon"]
df_daily = expand_to_daily(df, target_cols)
```

```python
# 4. 수질 데이터 merge
quality_cols = ["채수위치", "조사일"]

for col in df.columns:
    if any(x in col for x in ["수온", "pH", "DO", "탁도", "Chl-a"]):
        quality_cols.append(col)

df_daily = df_daily.merge(df[quality_cols], on=["채수위치", "조사일"], how="left")

# 5. 수질 보간
quality_features = [col for col in df_daily.columns if any(x in col for x in ["수온", "pH", "DO", "탁도", "Chl-a"])]

df_daily = df_daily.sort_values(["채수위치", "조사일"])

df_daily[quality_features] = (
    df_daily
    .groupby("채수위치")[quality_features]
    .transform(lambda x: x.interpolate().bfill().ffill())
)
```

```python
# 6. 발령단계
def classify_alert(x):
    if x >= 1_000_000:
        return "조류대발생"
    elif x >= 10_000:
        return "경계"
    elif x >= 1_000:
        return "관심"
    else:
        return "미발령"

df_daily["발령단계"] = df_daily["total_cyano"].apply(classify_alert)
```

```python
# 칼럼정리
df_daily["수온(℃)"] = df_daily["수온(℃)_y"]
df_daily["pH"] = df_daily["pH_y"]
df_daily["DO(㎎/L)"] = df_daily["DO(㎎/L)_y"]
df_daily["탁도"] = df_daily["탁도_y"]
df_daily["Chl-a (㎎/㎥)"] = df_daily["Chl-a (㎎/㎥)_y"]

df_daily = df_daily.drop(columns=[
    "수온(℃)_x", "수온(℃)_y",
    "pH_x", "pH_y",
    "DO(㎎/L)_x", "DO(㎎/L)_y",
    "탁도_x", "탁도_y",
    "Chl-a (㎎/㎥)_x", "Chl-a (㎎/㎥)_y"
], errors="ignore")

cols = ["수온(℃)", "pH", "DO(㎎/L)", "탁도", "Chl-a (㎎/㎥)"]

df_daily[cols] = (
    df_daily
    .groupby("채수위치")[cols]
    .transform(lambda x: x.interpolate().bfill().ffill())
)
```
